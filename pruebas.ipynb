# Elementos del Clasificador Gamma

* Operador Alfa
* Operador Beta
* Operador muBeta
* Codigo Johnson-Morbius Modificado

### Operador Alfa


```python
def op_Alfa(x_val, y_val):
    res = []
    for x, y in zip(x_val, y_val):
        if x == 0 and y == 0:
            res.append(1)
        elif x == 0 and y == 1:
            res.append(0)
        elif x == 1 and y == 0:
            res.append(2)
        elif x == 1 and y == 1:
            res.append(1)
    return res
```

### Operador Beta


```python
def op_Beta(x_val, y_val):
    res = []
    for x, y in zip(x_val, y_val):
        if x == 0 and y == 0:
            res.append(0)
        elif x == 0 and y == 1:
            res.append(0)
        elif x == 1 and y == 0:
            res.append(0)
        elif x == 1 and y == 1:
            res.append(1)
        elif x == 2 and y == 0:
            res.append(1)
        elif x == 2 and y == 1:
            res.append(1)
    return res
```

### Operador muBeta


```python
def op_muBeta(x):
    return sum(x)
```

### Módulo 2

```python
def mod_2(x_val):
    return [x % 2 for x in x_val]
```

```python
x = [0, 0, 0, 1, 0, 1, 1, 1]
y = [0, 1, 1, 0, 1, 0, 0, 1]

op_1 = op_Alfa(x, y)
print(op_1)

op_2 = mod_2(op_1)
print(op_2)

po_3 = op_muBeta(op_2)
print(po_3)
```
